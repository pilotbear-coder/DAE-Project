## t-SNE

```python
from sklearn import datasets
from sklearn.manifold import TSNE

from ipywidgets import interact, IntSlider


# Define a function that updates the t-SNE visualization based on the given parameters
def plot_tsne(perplexity=30, n_iter=1000):
    # Initialize the t-SNE object with specified perplexity, number of iterations, and a fixed random state for reproducibility
    tsne = TSNE(n_components=2, perplexity=perplexity, n_iter=n_iter, random_state=42)
    
    # Perform the t-SNE dimensionality reduction
    X_tsne = tsne.fit_transform(X)
    
    # Set up the plot
    plt.figure(figsize=(8, 6))
    # Create a scatter plot of the two t-SNE dimensions, color-coded by the iris species
    scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='viridis', edgecolor='k', alpha=0.7)
    # Add a color bar to the plot to show which colors correspond to which species
    plt.colorbar(scatter, label='Species')
    # Set the plot title, including dynamic parts for perplexity and iterations
    plt.title(f't-SNE visualization of the Iris dataset\nPerplexity: {perplexity}, Iterations: {n_iter}')
    # Label the x and y axes
    plt.xlabel('t-SNE feature 1')
    plt.ylabel('t-SNE feature 2')
    plt.grid(True)  # Enable a grid for easier visualization
    plt.show()  # Display the plot

# Create interactive widgets for the perplexity and number of iterations
interact(plot_tsne, perplexity=IntSlider(min=5, max=50, step=5, value=30), n_iter=IntSlider(min=250, max=5000, step=250, value=1000))
```


- **Import Required Libraries:**
  - `from sklearn.manifold import TSNE`: Import necessary components from sklearn for performing t-SNE.
  - `from ipywidgets import interact, IntSlider`: Import widgets for creating interactive controls in Jupyter Notebooks.


- **Define Interactive Visualization Function:**
  - `def plot_tsne(perplexity=30, n_iter=1000)`: Defines a function to perform t-SNE and plot the results, taking `perplexity` and `n_iter` as parameters to explore how these affect the outcome.
    - `tsne = TSNE(...)`: Initializes a t-SNE object with specified parameters.
    - `X_tsne = tsne.fit_transform(X)`: Applies t-SNE to reduce the dimensionality of the dataset to two components.
    - `plt.figure(figsize=(8, 6))`: Sets up a figure for plotting.
    - `scatter = plt.scatter(...)`: Creates a scatter plot, using a colormap and edge colors to enhance visibility.
    - `plt.colorbar(...)`: Adds a color bar to the plot for identifying species.
    - `plt.title(...)`, `plt.xlabel(...)`, `plt.ylabel(...)`: Adds a title and labels to axes with dynamic content reflecting current parameter settings.
    - `plt.grid(True)`: Enables a grid for better readability of the plot.
    - `plt.show()`: Displays the plot.

- **Set Up Interactive Widgets:**
  - `interact(plot_tsne, ...)`: Sets up interactive widgets with sliders for `perplexity` and `n_iter`, allowing the user to dynamically change these parameters and immediately see the updated t-SNE visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler

# Load the Iris dataset from sklearn's datasets
iris = datasets.load_iris()
X = iris.data  # The feature matrix (sepal length, sepal width, petal length, petal width)
y = iris.target  # The target variable (species of each iris flower)
species_names = iris.target_names  # Names of iris species for legend

# Standardize the feature data to have mean 0 and variance 1
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Define a range of perplexity values to explore t-SNE's sensitivity to this hyperparameter
perplexities = np.arange(5, 55, 5)

# Setup the subplot grid with 2 rows and 5 columns
n_rows = 2
n_cols = 5
fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 8))
axes = axes.flatten()  # Convert the matrix of axes into a flat array for easier iteration

# Color mapping for species - one color per species
colors = ['red', 'green', 'blue']

# Loop over each perplexity value and corresponding subplot axis
for ax, perplexity in zip(axes, perplexities):
    # Apply t-SNE with the current perplexity value
    tsne = TSNE(n_components=2, perplexity=perplexity, random_state=42)
    X_tsne = tsne.fit_transform(X_scaled)
    
    # Plot the t-SNE output for each species
    for i, color in enumerate(colors):
        mask = y == i  # Create a boolean mask for each species
        ax.scatter(X_tsne[mask, 0], X_tsne[mask, 1], color=color, label=species_names[i], edgecolor='k', alpha=0.6)
    
    # Configure the subplot
    ax.set_title(f'Perplexity: {perplexity}')  # Title showing the perplexity value
    ax.legend(title="Species")  # Add a legend with species names
    ax.grid(True)  # Enable grid for better visualization

# Adjust layout for neatness and display the plot
plt.tight_layout()
plt.show()
```

#### Import Required Libraries:
- **numpy, matplotlib.pyplot**: For numerical operations and data visualization.
- **datasets from sklearn**: To access pre-loaded datasets.
- **TSNE from sklearn.manifold**: To perform t-Distributed Stochastic Neighbor Embedding for dimensionality reduction.
- **StandardScaler from sklearn.preprocessing**: To normalize the data, ensuring features contribute equally.

#### Load and Prepare Data:
- **`iris = datasets.load_iris()`**: Loads the famous Iris flower dataset.
- **`X, y, species_names`**: Extracts feature data, species labels, and species names for use in plotting.

#### Data Standardization:
- **`StandardScaler()`**: Initializes a scaler that normalizes data to zero mean and unit variance, a prerequisite for many machine learning algorithms including t-SNE.

#### Define Perplexity Range and Setup Subplots:
- **`perplexities = np.arange(5, 55, 5)`**: Sets a range of perplexities to explore how this parameter affects the clustering.
- **`fig, axes = plt.subplots(n_rows, n_cols, figsize=(20, 8))`**: Creates a grid of subplots with specified dimensions.

#### t-SNE Application and Plotting:
- **Loop through `perplexities` and `axes.flatten()`**:
  - Apply t-SNE and plot results for each species using different colors to facilitate visual distinction.
  - **`tsne = TSNE(...)`**: Initializes and fits t-SNE for each perplexity setting.
  - **Scatter plots**: Different species are plotted using unique colors, with legends added for clarity.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

# Load the Iris dataset
data = load_iris()
X = data.data
y = data.target
species = data.target_names

# Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Apply t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity= 15)
X_tsne = tsne.fit_transform(X_scaled)

# Create a plot to compare PCA and t-SNE
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Colors for the different species
colors = ['r', 'g', 'b']

# Plotting the PCA results
for i, color in zip(range(len(species)), colors):
    ax1.scatter(X_pca[y == i, 0], X_pca[y == i, 1], color=color, label=species[i], edgecolor='k', alpha=0.7)
ax1.set_title('PCA of Iris Dataset')
ax1.set_xlabel('Principal Component 1')
ax1.set_ylabel('Principal Component 2')
ax1.legend(title="Species")

# Plotting the t-SNE results
for i, color in zip(range(len(species)), colors):
    ax2.scatter(X_tsne[y == i, 0], X_tsne[y == i, 1], color=color, label=species[i], edgecolor='k', alpha=0.7)
ax2.set_title('t-SNE of Iris Dataset')
ax2.set_xlabel('t-SNE Feature 1')
ax2.set_ylabel('t-SNE Feature 2')
ax2.legend(title="Species")

# Display the plot
plt.show()
```

#### Import Libraries
- **numpy, matplotlib.pyplot**: For numerical operations and plotting.
- **PCA, TSNE from sklearn**: To perform Principal Component Analysis and t-Distributed Stochastic Neighbor Embedding.
- **load_iris, StandardScaler from sklearn**: For loading the Iris dataset and standardizing the features.

#### Load and Prepare Data
- `data = load_iris()`: Load the Iris dataset which includes measurements of iris plants and their species classification.
- `X, y, species`: Extract features, target labels, and species names for later use in plots.

#### Standardize the Data
- `StandardScaler()`: Ensures features contribute equally by standardizing to mean 0 and variance 1, crucial for PCA.

#### Dimensionality Reduction
- `PCA(n_components=2)`: Reduce dimensions to 2 principal components for visualization.
- `TSNE(n_components=2, random_state=42)`: Apply t-SNE to project the data into a 2D space, set random state for reproducibility.

#### Create Plot
- Setup two subplots for PCA and t-SNE results. Use `subplots` with a fixed size.
- **Color Assignment**: Define color codes for the three species of Iris.

#### Plotting Results with Legends
- Loop over each species to plot PCA and t-SNE results:
  - Scatter plot for each species colored differently with labels.
  - `ax1.legend()`, `ax2.legend()`: Add legends to each subplot to identify species by color.

#### Display Plot
- `plt.show()`: Render the combined figure with PCA and t-SNE plots, each subplot having a legend indicating species.




### PCA Visualization Interpretation:

- The **PCA plot** shows the dataset reduced to two principal components, representing the axes of maximum variance.
- There are three groups visible, corresponding to the Iris flower species: *setosa*, *versicolor*, and *virginica*.
- *Setosa* (red) is well separated from the other two species along the first principal component, indicating distinct feature measurements.
- *Versicolor* (green) and *virginica* (blue) show some overlap, suggesting more similarity between these species, although they are mostly separable along the second principal component.

### t-SNE Visualization Interpretation:

- The **t-SNE plot** provides a view of the dataset where local similarities are preserved, often revealing clustering patterns not apparent in PCA.
- All three Iris species form distinct clusters, with *setosa* (red) showing a tight grouping, emphasizing the dissimilarity with the other species.
- *Versicolor* (green) and *virginica* (blue) also appear as distinct clusters with minimal overlap, indicating t-SNE's effectiveness in species differentiation based on local structure.

### Comparative Insights:

- PCA focuses on capturing the global structure and variance, sometimes missing the finer, nonlinear relationships, which can lead to overlaps in complex datasets.
- t-SNE excels at preserving local neighbor relations, revealing data clusters and separations even when PCA cannot, though it provides less interpretability in terms of the original feature space.
- The clarity of separation in t-SNE makes it particularly valuable for exploratory data analysis and understanding the inherent grouping within the data.

In summary, t-SNE provides better cluster separation for the Iris dataset, making it a superior choice for visualization when the goal is to distinguish between closely related groups. PCA, while offering a quicker and more interpretable reduction, might not always provide clear separability, especially when linear relationships do not dominate the data structure.
