```python
numeric_vars = ['Order_ID', 'Units_Sold', 'MRP', 'Discount_Applied', 'Revenue', 'Profit']


for var in numeric_vars:
    plt.figure(figsize=(8, 5))
    sns.kdeplot(data=df, x=var, fill=True, linewidth=2, color='skyblue')
    plt.title(f'KDE Plot for {var}', fontsize=14)
    plt.xlabel(var, fontsize=12)
    plt.ylabel('Density', fontsize=12)
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
gender_counts = df['Gender_Category'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightcoral', 'lightgreen'])
plt.title('Gender Category Distribution')
plt.axis('equal')  # Ensures pie is a circle
plt.tight_layout()
plt.show()
```

```python
channel_counts = df['Sales_Channel'].value_counts().reset_index()
channel_counts.columns = ['Sales_Channel', 'Count']

plt.figure(figsize=(7, 5))
sns.barplot(data=channel_counts, x='Sales_Channel', y='Count', palette='pastel')

plt.title('Sales Channel Frequency Distribution', fontsize=14)
plt.xlabel('Sales Channel', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

for index, row in channel_counts.iterrows():
    plt.text(index, row['Count'] + 5, f"{row['Count']:,.0f}", ha='center', fontsize=10)

plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
product_line_counts = df['Product_Line'].value_counts().reset_index()
product_line_counts.columns = ['Product_Line', 'Count']

plt.figure(figsize=(8, 5))
sns.barplot(data=product_line_counts, x='Product_Line', y='Count', palette='Blues')

plt.title('Frequency of Product Line', fontsize=14)
plt.xlabel('Product Line', fontsize=12)
plt.ylabel('Number of Entries', fontsize=12)

for index, row in product_line_counts.iterrows():
    plt.text(index, row['Count'] + 2, f"{row['Count']}", ha='center', fontsize=10)

plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
df_clean = df[['Product_Name', 'Units_Sold']].dropna()

top_products = df_clean.groupby('Product_Name')['Units_Sold'].sum().sort_values(ascending=False).head(10)

top_products = top_products.reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=top_products, y='Product_Name', x='Units_Sold', palette='coolwarm')

plt.title('Top 10 Best-Selling Products by Units Sold', fontsize=14)
plt.xlabel('Total Units Sold', fontsize=12)
plt.ylabel('Product Name', fontsize=12)


for i, row in top_products.iterrows():
    plt.text(row['Units_Sold'] + 1, i, f"{row['Units_Sold']:,.0f}", va='center', fontsize=10)

plt.tight_layout()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
df_clean = df.dropna(subset=['Product_Line', 'Product_Name'])

plt.figure(figsize=(12, 7))
sns.countplot(data=df_clean, x='Product_Line', hue='Product_Name', palette='tab10')

plt.title('Products Produced per Product Line', fontsize=14)
plt.xlabel('Product Line', fontsize=12)
plt.ylabel('Product Count', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Product Name', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
df_clean = df.dropna(subset=['Region', 'Gender_Category'])

plt.figure(figsize=(10, 6))
sns.countplot(data=df_clean, x='Region', hue='Gender_Category', palette='Set2')

plt.title('Gender Category Distribution Across Regions', fontsize=14)
plt.xlabel('Region', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend(title='Gender Category')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
