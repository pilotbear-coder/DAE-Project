```python
import plotly.express as px

df = px.data.iris()
df
```

```python
fig = px.scatter(df, x="sepal_length", y="sepal_width", facet_col="species")
# sources of images
sources = [
    "https://upload.wikimedia.org/wikipedia/commons/thumb/f/fe/Iris_setosa_var._setosa_%282595031014%29.jpg/360px-Iris_setosa_var._setosa_%282595031014%29.jpg",
    "https://upload.wikimedia.org/wikipedia/commons/thumb/3/38/Iris_versicolor_quebec_1.jpg/320px-Iris_versicolor_quebec_1.jpg",
    "https://upload.wikimedia.org/wikipedia/commons/thumb/f/f8/Iris_virginica_2.jpg/480px-Iris_virginica_2.jpg",
    #northeastern logo from https://brand.northeastern.edu/visual-identity/seal/
    # "https://brand.northeastern.edu/wp-content/uploads/2022/06/seal-black.svg"

]
# add images  #enumerate documentation https://docs.python.org/3/library/functions.html
for col, src in enumerate(sources):
    fig.add_layout_image(
        row=1,
        col=col + 1,
        source=src,
        xref="x domain",
        yref="y domain",
        x=1,
        y=1,
        xanchor="right",
        yanchor="top",
        sizex=0.2,
        sizey=0.2,
        #opacity=0.3,
        #layer="below",
        #sizing="stretch"
    )

fig.show()
```

```python
from dash import Dash, html
import plotly.express as px
import pandas as pd
from dash import dcc

app = Dash(__name__, suppress_callback_exceptions=True)

df = pd.DataFrame({
    "Fruit": ["Apples", "Oranges", "Bananas", "Apples", "Oranges", "Bananas"],
    "Amount": [4, 1, 2, 2, 4, 5],
    "City": ["SF", "SF", "SF", "Montreal", "Montreal", "Montreal"]
})

fig = px.bar(df, x="Fruit", y="Amount", color="City", barmode="group")

# df_1 = px.data.iris()  # iris pandas DataFrame
# fig_1 = px.scatter(df_1, x="sepal_width", y="sepal_length")

app.layout = html.Div(children=[
    html.H1(children='Hello Dash'),
    html.Div(children='Hello World'),
    dcc.Dropdown(['New York City', 'Montréal', 'San Francisco'], 'Montréal'),
    html.Div([
    dcc.Graph(
        id='graph1',
        figure=fig,
        #style={'display': 'inline-block'}
    ),
    dcc.Graph(
        id='graph2',
        figure=fig,
        style={'display': 'inline-block'}
    ),
    dcc.Graph(
        id='graph3',
        figure=fig,
        style={'display': 'inline-block'}
    ),
    ],
    )

    # dcc.Graph(
    #     id='graph',
    #     figure=fig_1
    # )
])

if __name__ == '__main__':
    app.run(jupyter_mode="external",
            #debug='True'
            )
```
