```python
class MyLinearRegression:

    # initialization function for the class
    def __init__(self, X, y, max_iteration, epsilon, learning_rate, gd) -> None:
        self.X = X
        self.y = y
        self.max_iteration = max_iteration
        self.epsilon = epsilon
        self.learning_rate = learning_rate
        self.gd = gd

    # write a function to split the data into test and train sets
    # write a function to split the data into test and train sets
    def data_split(self, test_size=0.2, random_state=42):
        # Using the imported train_test_split
        X_train, X_test, y_train, y_test = train_test_split(
            self.X, 
            self.y, 
            test_size=test_size, 
            random_state=random_state
        )
        
        # Storing them as attributes so other methods can access them
        self.X_train = X_train
        self.X_test = X_test
        self.y_train = y_train
        self.y_test = y_test
        
        return X_train, X_test, y_train, y_test
        

    # write a function to add the bias column to the input
    def add_X0(self, X):
        if isinstance(X, pd.DataFrame):
            X = X.values
            
        # Create a column of ones with the same number of rows as X
        ones = np.ones((X.shape[0], 1))
        
        # Horizontal stack: [1s, X]
        X_with_bias = np.hstack((ones, X))
        
        return X_with_bias

    # write a function to scale the train data
    def train_scaling(self, X):
        self.mean = np.mean(X, axis=0)
        self.std = np.std(X, axis=0)
        
        # Prevent division by zero if a feature has zero variance
        self.std[self.std == 0] = 1
        
        # Standardize: (X - mean) / std
        X_scaled = (X - self.mean) / self.std
        
        return X_scaled
    
    # write a function to scale the test data
    def test_scaling(self, X, mean, std):
        if self.mean is None or self.std is None:
            raise ValueError("Training data must be scaled before scaling test data.")

        # Standardize: (X_test - training_mean) / training_std
        X_scaled = (X - self.mean) / self.std
        
        return X_scaled
    
    # the following functions are to check the rank of the input matrix
    def rank(self, X):
        v, s, u = np.linalg.svd(X)  # singular value decomposition
        rank = np.sum(s>0.00001)  # s is an array of eigen values and we need all non-zero eigen values

        rank = np.linalg.matrix_rank(X)
        return rank

    def full_rank(self,X):
        # minimum of n and d should be equal to rank for it to be a full rank
        rank = self.rank(X)
        if rank == min(X.shape):
          self.fullRank = True
          print("Data is Full Rank")
        else:
          self.fullRank = False
          print("Data is not Full Rank")

    def low_rank(self, X):
        if X.shape[0] < X.shape[1]: # n<d
          self.lowRank = True
          print("Data is low Rank")
        else:
          self.lowRank = False
          print("Data is not low Rank")

    # write a function for closed_form_solution
    def closed_form_solution(self, X, y):
        XT = X.T
        
        # Calculate (X^T * X)
        XTX = np.dot(XT, X)
        
        # Calculate the inverse. 
        # Using np.linalg.pinv (Moore-Penrose pseudo-inverse) is safer than 
        # np.linalg.inv because it handles non-full rank (singular) matrices,
        # which you previously checked for.
        XTX_inv = np.linalg.pinv(XTX)
        
        # Calculate X^T * y
        XTy = np.dot(XT, y)
        
        # Calculate w = (X^T * X)^-1 * X^T * y
        self.w = np.dot(XTX_inv, XTy)
        
        return self.w

    # write the predict function
    def predict(self, X):
        """
        Predicts the target values for the given input features X.
        Note: X must have the bias column (X0) added and should be scaled 
        similarly to the training data.
        """
        # Ensure the weights have been calculated
        if not hasattr(self, 'w'):
            raise ValueError("The model has not been trained yet. Please run a solver (e.g., closed_form_solution) first.")

        # If X is a DataFrame, convert to numpy for the dot product
        if isinstance(X, pd.DataFrame):
            X = X.values

        # Perform matrix multiplication: y_pred = Xw
        y_pred = np.dot(X, self.w)

        return y_pred

    # write the functions to calculate root mean squared error and sum of squared errors to evaluate the models and create the cost functions
    def sse(self, X, y):
        """
        Calculates the Sum of Squared Errors (SSE).
        This is the raw sum of the squared differences between actual and predicted.
        """
        y_hat = self.predict(X)
        # sse = sum((y_actual - y_predicted)^2)
        return np.sum((y - y_hat) ** 2)

    # write the function to calculate root mean squared error
    # write a function to calculate sum of squared errors
    def sse(self, X, y):
        """
        Calculates the Sum of Squared Errors (SSE).
        This is the raw sum of the squared differences between actual and predicted.
        """
        y_hat = self.predict(X)
        # sse = sum((y_actual - y_predicted)^2)
        return np.sum((y - y_hat) ** 2)

    # write the function to calculate root mean squared error
    def rmse(self, X, y):
        """
        Calculates the Root Mean Squared Error (RMSE).
        This provides the average error magnitude in the same units as the target variable.
        """
        y_hat = self.predict(X)
        # mse = mean((y_actual - y_predicted)^2)
        mse = np.mean((y - y_hat) ** 2)
        
        return np.sqrt(mse)

    # the following two functions (cost_funtion and cost_derivative) are cost functions that will be used in the gradient descent function
    def cost_function(self, X, y):
        loss = self.rmse(X,y)
        return loss / 2

    def cost_derivative(self, X, y):
        y_hat = self.predict(X)
        return X.T.dot(y_hat - y)

    # write a function to perform gradient descent
    def gradient_descent(self, X, y):
        # The number of weights equals the number of features (including the bias column)
        self.w = np.zeros(X.shape[1])
        
        # Array to store the cost at each iteration to track progress
        self.cost_history = []

        # 2. Start the optimization loop
        # We use tqdm for a progress bar as it was imported in your environment
        for i in tqdm(range(self.max_iteration)):
            
            # Save old weights to check for convergence later
            old_w = self.w.copy()
            
            # 3. Calculate the gradient using your cost_derivative function
            # Gradient = X.T.dot(y_hat - y)
            gradient = self.cost_derivative(X, y)
            
            # 4. Update weights: w = w - (alpha * gradient)
            self.w = self.w - (self.learning_rate * gradient)
            
            # 5. Calculate current cost and save it to history
            current_cost = self.cost_function(X, y)
            self.cost_history.append(current_cost)
            
            # 6. Check for convergence
            # If the change in the weight vector is smaller than epsilon, stop early
            if np.linalg.norm(self.w - old_w) < self.epsilon:
                print(f"Gradient Descent converged at iteration {i}")
                break
                
        return self.w

    # write the fit function
    def fit(self):
        # 1. Split the data into train and validation sets
        X_train, X_val, y_train, y_val = self.data_split()

        # 2. Scale the training data 
        # (We convert to numpy arrays to ensure math operations work smoothly)
        X_train_scaled = self.train_scaling(X_train.values if hasattr(X_train, 'values') else X_train)
        
        # 3. Add the bias column (X0) to the scaled training data
        X_train_final = self.add_X0(X_train_scaled)

        # 4. Choose the solver based on the 'gd' flag
        if self.gd:
            print("Training with Gradient Descent...")
            self.w = self.gradient_descent(X_train_final, y_train.values if hasattr(y_train, 'values') else y_train)
        else:
            print("Training with Closed Form Solution...")
            self.w = self.closed_form_solution(X_train_final, y_train.values if hasattr(y_train, 'values') else y_train)

        # 5. Calculate and report training performance
        train_rmse = self.rmse(X_train_final, y_train)
        print(f"Training Complete. Training RMSE: {train_rmse:.4f}")

        return self
```

```python
class MyLinearRegression:

    # initialization function for the class
    def __init__(self, X, y, max_iteration, epsilon, learning_rate, gd,reg,reg_param) -> None:
        self.X = X
        self.y = y
        self.max_iteration = max_iteration
        self.epsilon = epsilon
        self.learning_rate = learning_rate
        self.gd = gd
        self.reg=reg
        self.reg_param=reg_param

    # write a function to split the data into test and train sets
    def data_split(self):
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, 
            self.y, 
            test_size=0.20, 
            random_state=42
        )

    # write a function to add the bias column to the input
    def add_X0(self, X):
        X_with_bias = np.column_stack([np.ones(X.shape[0]), X])
        return X_with_bias

    # write a function to scale the train data
    def train_scaling(self, X):
        self.mean = np.mean(X, axis=0)
        self.std = np.std(X, axis=0)

    # write a function to scale the test data
    def test_scaling(self, X, mean, std):
        X_scaled = (X - mean) / std
        
        return X_scaled

    # the following functions are to check the rank of the input matrix
    def rank(self, X):
        v, s, u = np.linalg.svd(X)  # singular value decomposition
        rank = np.sum(s>0.00001)  # s is an array of eigen values and we need all non-zero eigen values

        rank = np.linalg.matrix_rank(X)
        return rank

    def full_rank(self,X):
        # minimum of n and d should be equal to rank for it to be a full rank
        rank = self.rank(X)
        if rank == min(X.shape):
          self.fullRank = True
          print("Data is Full Rank")
        else:
          self.fullRank = False
          print("Data is not Full Rank")

    def low_rank(self, X):
        if X.shape[0] < X.shape[1]: # n<d
          self.lowRank = True
          print("Data is low Rank")
        else:
          self.lowRank = False
          print("Data is not low Rank")

    # write a function for closed_form_solution
    def closed_form_solution(self, X, y):
        n_features = X.shape[1]
        
        # Calculate X^T * X
        XTX = X.T @ X
        
        if self.reg:
            # Create the Identity matrix for regularization
            # We typically do not regularize the intercept (first column)
            I = np.eye(n_features)
            I[0, 0] = 0 
            
            # Add the regularization term: (X^T*X + lambda*I)
            # This ensures the matrix is invertible even if it's not full rank
            A = XTX + self.reg_param * I
        else:
            A = XTX

        # Calculate X^T * y
        b = X.T @ y

        # Solve for W. We use pinv (pseudo-inverse) for maximum numerical stability
        # especially if the matrix is low-rank or singular.
        self.w = np.linalg.pinv(A) @ b
        
        return self.w

    # write the predict function
    def predict(self, X):
        X_scaled = self.test_scaling(X, self.mean, self.std)
        
        # 2. Add the bias column (X0 = 1)
        X_final = self.add_X0(X_scaled)
        
        # 3. Calculate predictions using the dot product
        # y_hat = X * w
        y_hat = X_final @ self.w
        
        return y_hat

    # write the functions to calculate root mean squared error and sum of squared errors to evaluate the models and create the cost functions
    def rmse(self, X, y):
        mse = self.sse(X, y) / len(y)
        
        # Square root of MSE gives RMSE
        return np.sqrt(mse)

    def sse(self,X,y):
        y_hat = self.predict(X)
        
        # 2. Calculate residuals (errors)
        error = y_hat - y
        
        # 3. Square the errors and sum them up
        # We can use np.dot(error, error) as a shorthand for sum(error^2)
        return np.sum(np.square(error))

    # the following two functions (cost_funtion and cost_derivative) are cost functions that will be used in the gradient descent function
    def cost_function(self, X, y):
        loss = self.rmse(X,y)
        return loss / 2

    def cost_derivative(self, X, y):
        n = len(y)
        # 1. Get predictions: y_hat = X * w
        # Note: In the GD loop, X here will already be scaled with bias added
        y_hat = X @ self.w

    # write a function to perform gradient descent
    def gradient_descent(self, X, y):
        n_features = X.shape[1]
        self.w = np.zeros(n_features)
        
        # Track cost history to monitor convergence
        self.cost_history = []

        for i in range(self.max_iteration):
            # 2. Calculate the gradient (slope) at the current weights
            gradient = self.cost_derivative(X, y)
            
            # 3. Update weights: w = w - (learning_rate * gradient)
            new_w = self.w - (self.learning_rate * gradient)
            
            # 4. Check for convergence (stopping criteria)
            # If the change in weights is smaller than epsilon, we stop early
            if np.linalg.norm(new_w - self.w) < self.epsilon:
                print(f"Gradient Descent converged at iteration {i}")
                break
            
            # 5. Update the weight attribute and record the cost
            self.w = new_w
            current_cost = self.cost_function(X, y)
            self.cost_history.append(current_cost)
            
            # Optional: Print progress every 1000 iterations
            if i % 1000 == 0:
                print(f"Iteration {i}: Cost = {current_cost:.4f}")

        return self.w

    # write the fit function
    def fit(self):
        self.data_split()

        # 2. Perform Rank checks to understand data stability
        self.full_rank(self.X_train)
        self.low_rank(self.X_train)

        # 3. Scale the training data and save the mean/std for future predictions
        X_train_scaled = self.train_scaling(self.X_train)

        # 4. Scale the test data using the training parameters
        X_test_scaled = self.test_scaling(self.X_test, self.mean, self.std)

        # 5. Add the bias column (X0) to both sets
        # These are the versions used for internal math
        self.X_train_final = self.add_X0(X_train_scaled)
        self.X_test_final = self.add_X0(X_test_scaled)

        # 6. Choose the solver based on the user's initialization
        if self.gd:
            print("Training with Gradient Descent...")
            self.w = self.gradient_descent(self.X_train_final, self.y_train)
        else:
            print("Training with Closed Form Solution (Normal Equation)...")
            self.w = self.closed_form_solution(self.X_train_final, self.y_train)
```
