 results in a probability of zero. This occurs because the feature pathetic=1 never appears in any "positive" samples in the training data (p(pathetic=1∣positive)=0/4). Since Naive Bayes multiplies all feature probabilities together, a single zero value zeros out the entire product, regardless of how strong the other evidence (like fun=1) might be.

To deal with this problem, we would use Laplace Smoothing

# QUESTION 2: WRITING A NAIVE BAYES CLASSIFIER FROM SCRATCH

Write a class to perform discrete Naive Bayes for a random dataset with n = 1000 (# of data points) and d = 2 (# of dimensions) for a binary classification y [0,1]. x1 and x2 can get any values between [0 and 1]. Then, implement the Laplace Smoothing function for the Naive Bayes class

### Import the necessary libraries

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from tqdm import tqdm
from scipy.stats import norm
from sklearn.model_selection import train_test_split
```

### Creating a sample dataset with two dimensions x1, x2

```python
x1 = np.random.randint(2, size = 1000)
x2 = np.random.randint(2, size = 1000)

y= np.random.randint(2, size = 1000)
```

### Create an array to store the discrete generated variables x1 and x2

```python
X = np.column_stack((x1, x2))

print(f"Shape of feature matrix X: {X.shape}")
print(f"First 5 rows:\n{X[:5]}")
```

### Create the class for Naive Bayes Classifier

{5 marks}

```python
class Naive_Bayes:

    def __init__(self, X, y, laplace=0):
        self.X = X
        self.y = y
        self.laplace = laplace
        self.priors = {}
        self.likelihoods = {}

    def trainTestSplit(self):
        # Splitting the data into training and testing sets (80/20)
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=0.2, random_state=42
        )

    def runModel(self):
        # Safeguard: Ensure data is split before training
        if not hasattr(self, 'X_train'):
            self.trainTestSplit()

        # Subsets for class 0 and 1
        X_train_0 = self.X_train[self.y_train == 0]
        X_train_1 = self.X_train[self.y_train == 1]

        # Priors: P(Y=0) and P(Y=1)
        self.priors[0] = len(X_train_0) / len(self.y_train)
        self.priors[1] = len(X_train_1) / len(self.y_train)

        # Likelihoods for binary features (k=2)
        k = self.X_train.shape[1]
        for c, subset in enumerate([X_train_0, X_train_1]):
            self.likelihoods[c] = {}
            for i in range(k):
                # Count instances where feature i == 1
                num_ones = np.sum(subset[:, i] == 1)
                # Apply Laplace smoothing formula
                prob_1 = (num_ones + self.laplace) / (len(subset) + self.laplace * 2)
                self.likelihoods[c][i] = prob_1

    def predict(self):
        # Safeguard: Ensure model is trained before predicting
        if not self.likelihoods:
            self.runModel()

        predictions = []
        for sample in self.X_test:
            posteriors = []
            for c in [0, 1]:
                # Start with the prior
                p_class = self.priors[c]
                # Multiply by conditional probabilities (Likelihoods)
                for i, val in enumerate(sample):
                    p_feat_1 = self.likelihoods[c][i]
                    # If feature is 1, use p_feat_1; if 0, use (1 - p_feat_1)
                    p_class *= p_feat_1 if val == 1 else (1 - p_feat_1)
                posteriors.append(p_class)
            
            # Argmax: index 0 or 1
            predictions.append(np.argmax(posteriors))
            
        return np.array(predictions)
```
